```python
import pandas as pd
import numpy as np
from itertools import combinations
```

```python
url = "https://raw.githubusercontent.com/darkeyr0728/Enzor_Reyes_1714662012/refs/heads/main/data/clave_E_agrupacion.csv"

df = pd.read_csv(url)

df.head()
```

```python
df.shape
df.columns
df.info()
```

```python
# 3. Columnas numéricas para análisis
columnas_numericas = [
    "edad",
    "ingresos",
    "frecuencia_uso",
    "gasto_promedio",
    "satisfaccion",
    "reclamos",
    "antiguedad_meses"
]

```

```python
# 4. Manejo de valores nulos
# Se reemplazan nulos con la mediana
df[columnas_numericas] = df[columnas_numericas].fillna(
    df[columnas_numericas].median()
)
```

```python
# 5. Valores atípicos por media +/- 2 desviaciones estándar
for columna in columnas_numericas:
    media = df[columna].mean()
    desviacion = df[columna].std()

    limite_inferior = media - 2 * desviacion
    limite_superior = media + 2 * desviacion

    atipicos = df[
        (df[columna] < limite_inferior) |
        (df[columna] > limite_superior)
    ]

    print(
        columna,
        "Límite inferior:", limite_inferior,
        "Límite superior:", limite_superior,
        "Cantidad atípicos:", len(atipicos)
    )

```

```python
# 6. Variables del modelo
X = df[columnas_numericas]
```

```python
from sklearn.preprocessing import StandardScaler

# 7. Escalamiento de datos
# Se usa porque ingresos, gasto_promedio, reclamos, edad, etc.
# tienen escalas diferentes
scaler = StandardScaler()
X_escalado = scaler.fit_transform(X)
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# 8. K-Means con K=2
modelo_k2 = KMeans(n_clusters=2, random_state=0, n_init=10)
df["cluster_k2"] = modelo_k2.fit_predict(X_escalado)

print("\nCantidad por clúster K=2:")
print(df["cluster_k2"].value_counts())

print("\nCentroides K=2:")
centroides_k2 = df.groupby("cluster_k2")[columnas_numericas].mean()
print(centroides_k2)

print("Silhouette K=2:", silhouette_score(X_escalado, df["cluster_k2"]))
```

```python
# 9. K-Means con K=4
modelo_k4 = KMeans(n_clusters=4, random_state=0, n_init=10)
df["cluster_k4"] = modelo_k4.fit_predict(X_escalado)

print("\nCantidad por clúster K=4:")
print(df["cluster_k4"].value_counts())

print("\nCentroides K=4:")
centroides_k4 = df.groupby("cluster_k4")[columnas_numericas].mean()
print(centroides_k4)

print("Silhouette K=4:", silhouette_score(X_escalado, df["cluster_k4"]))
```

```python
# 10. Identificar clúster de mayor riesgo
# Mayor riesgo: más reclamos y menor satisfacción
centroides_k4["riesgo"] = (
    centroides_k4["reclamos"] -
    centroides_k4["satisfaccion"]
)

cluster_riesgo = centroides_k4["riesgo"].idxmax()

print("\nClúster de mayor riesgo:", cluster_riesgo)
print("\nDatos promedio del clúster de mayor riesgo:")
print(centroides_k4.loc[cluster_riesgo])

```

```python
# 11. Filtrar personas del clúster de riesgo
personas_riesgo = df[df["cluster_k4"] == cluster_riesgo]

print("\nResumen de personas en riesgo:")
print(personas_riesgo[columnas_numericas].describe())
```

```python
# 12. Exportar resultado
personas_riesgo.to_csv(
    "personas_riesgo_cluster_kmeans.csv",
    index=False
)

print("\nArchivo generado: personas_riesgo_cluster_kmeans.csv")
```
