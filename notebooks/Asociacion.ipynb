```python
import pandas as pd
import numpy as np
from itertools import combinations
```

```python
url = "https://raw.githubusercontent.com/darkeyr0728/Enzor_Reyes_1714662012/refs/heads/main/data/clave_E_asociacion.csv"

df = pd.read_csv(url)

df.head()
```

```python
df.shape
df.columns
df.info()
```

```python
df.isnull().sum()
df.duplicated().sum()
```

```python
media = df["cantidad"].mean()
desviacion = df["cantidad"].std()
limite_inferior = media - 2 * desviacion
limite_superior = media + 2 * desviacion
outliers = df[(df["cantidad"] < limite_inferior) | (df["cantidad"] > limite_superior)]
print("cantidad", limite_inferior, limite_superior, len(outliers))
```

```python
# Agrupamos por transacción y cliente para obtener los ítems comprados juntos
df_asociacion = df.groupby("transaccion_id")["item"].apply(list)
```

```python
# Convertimos a formato booleano (matriz de transacciones vs ítems)
todos_items = sorted(set(df["item"]))
df_asociacion_bool = pd.DataFrame([
    {item: (item in items) for item in todos_items}
    for items in df_asociacion
])
```

```python
from mlxtend.frequent_patterns import apriori, association_rules


# El parámetro 'min_support' es el umbral de soporte mínimo 0.10
frequent_itemsets = apriori(df_asociacion_bool, min_support=0.10, use_colnames=True)


# El parámetro 'metric' define cómo evaluar las reglas confianza
# El parámetro 'min_threshold' es el umbral de confianza mínimo 0.50
reglas_df_mlxtend = association_rules(frequent_itemsets, metric="confianza", min_threshold=0.50)

# Ordenar las reglas
reglas_df_mlxtend = reglas_df_mlxtend.sort_values(by=["confianza", "soporte"], ascending=False)

print(reglas_df_mlxtend)
```

```python
# 6. Guardar resultados
df_asociacion_bool.to_csv("transacciones_preparado_asociacion.csv", index=False)
reglas_df_mlxtend.to_csv("transacciones_reglas_asociacion_mlxtend.csv", index=False)
```
